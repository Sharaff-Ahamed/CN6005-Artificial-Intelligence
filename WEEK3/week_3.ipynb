```python
import pandas as pd
import os
import urllib.request

col_names = ['pregnant', 'glucose', 'bp', 'skin', 'insulin', 'bmi', 'pedigree', 'age', 'label']
csvfile = "diabetesN.csv"
csvfilePath = '/content/drive/MyDrive/UELAI/' + csvfile

# Check if the Google Drive file exists; if not, download a fallback copy
if not os.path.exists(csvfilePath):
    print(f"File not found at {csvfilePath}. Downloading a public copy of the dataset instead...")
    fallback_url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"
    csvfilePath = csvfile
    if not os.path.exists(csvfilePath):
        urllib.request.urlretrieve(fallback_url, csvfilePath)

# Load dataset
pima = pd.read_csv(csvfilePath, header=None, names=col_names)
pima.head()
```
